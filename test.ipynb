```python
import asyncio
import time
```

```python
def say_after_(delay, what):
    time.sleep(delay)
    print(what)

say_after_(2, "один")
say_after_(2, "два")
```

```python
async def say_after(delay, what):
    await asyncio.sleep(delay)
    print(what)

async def main():
    await say_after(2, "один")
    await say_after(2, "два")

await main()   # 👈 вместо asyncio.run(main())
```

```python
async def main():
    await asyncio.gather(
        say_after(1, "один"),
        say_after(1, "два"),
    )
```

```python
await main()
```

```python
async def main():
    task = asyncio.create_task(say_after(2, "в фоне"))
    print("я не жду задачу!")
    await task   # а вот теперь дождусь
```

```python
await main()
```

```python
import asyncio

async def task_a():
    print("A: начало")
    await asyncio.sleep(4)         # ← A встаёт здесь на 2 секунды
    print("A: конец")              # ← эта строка ждёт

async def task_b():
    print("B: начало")
    await asyncio.sleep(3)
    print("B: конец")

async def main():
    await asyncio.gather(task_a(), task_b())   # запускаем ОБЕ параллельно

await main()
```
